```python
import os
import json
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
```

```python
#
# OPENAI offers this endpoint: "https://api.openai.com/v1/chat/completions",
# to interact with the models onlline
#
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gpt-5-nano",
    "messages": [
        {"role": "user", "content": "Tell me a fun fact"}]
}

response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

response.json()
```

```python
#
# Again we show again the content of the response
#
d = response.json()
print(f'Keys in the response message: {d.keys()}')
for k,v in response.json().items():
    if k not in ['choices', 'usage']:
        print(f' --- {k} : {v}')
    if k in ['usage']:
        print(' --- usage:')
        for usage_k,usage_v in v.items():
            print(f'      --- {usage_k} : {usage_v}')
    if k == 'choices':
        print(f' --- choices: {len(d['choices'])} ')
        for item in v:
            print(f'       - keys: {item.keys()}')          
            for item_k,item_v in item.items():
                if item_k == 'message':
                    print('            - message:')
                    for message_k,message_v in item_v.items():
                        print(f'              - {message_k} : {message_v}')
                else:    
                    print(f'         - {item_k} : {item_v}')
```

```python
##########################################
#
###  Using the API endpoint is a bit of a pain so we can usage the OpenAI package directly
#
###########################################
## in the end this is just a wrapper over the endpoint above
## and it's open source if you want to check the code

from openai import OpenAI
openai = OpenAI()

response = openai.chat.completions.create(model="gpt-5-nano", messages=[{"role": "user", "content": "Tell me a fun fact"}])

response.choices[0].message.content
```

```python
##########################################
#
###  Using the the same code we can query different model and in particular we can also use the locally running ollama
# Note that we assume that
#  - ollama is running locally as a container (and port 11434 is open)
#  - the key is mandatory but you can just use ollama
#  - you need to specify a model you downloaded
#
ollama_base_api = "http://localhost:11434/v1"
ollama = OpenAI(base_url=ollama_base_api, api_key='ollama')

response = ollama.chat.completions.create(model="llama3:latest", 
                                          messages=[{"role": "user", "content": "Tell me a fun fact"}]
                                          )

response.choices[0].message.content
```
